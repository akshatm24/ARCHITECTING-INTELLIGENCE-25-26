````python
import re

def clean_text(text: str) -> str:
    # Convert weird non-breaking spaces to normal spaces
    text = text.replace("\xa0", " ")

    # Remove hyphenation from line breaks: "retriev-\nal" -> "retrieval"
    text = re.sub(r"(\w)-\n(\w)", r"\1\2", text)

    # Replace single newlines inside paragraphs with spaces
    # Keep paragraph breaks (2+ newlines) as paragraph breaks
    text = re.sub(r"(?<!\n)\n(?!\n)", " ", text)

    # Collapse multiple spaces
    text = re.sub(r"[ \t]+", " ", text)

    # Collapse 3+ newlines into 2 newlines
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()
````

````python
clean_pages_text = [clean_text(p) for p in pages_text]
print("Cleaned pages:", len(clean_pages_text))

print("\n===== CLEAN PAGE 1 PREVIEW =====\n")
print(clean_pages_text[0][:1200])
````

````python
clean_full_text = "\n\n".join(clean_pages_text)

print("Clean full length:", len(clean_full_text))
print("\n===== CLEAN DOC PREVIEW =====\n")
print(clean_full_text[:1500])
````

````python
import chromadb
````

````python
chroma_client = chromadb.Client()
````

````python
collection = chroma_client.create_collection(
    name="rag_research_paper"
)
````

````python
ids = []
documents = []
embeddings_list = []

for item in chunk_embeddings:
    ids.append(item["id"])
    documents.append(item["text"])
    embeddings_list.append(item["embedding"])
````

````python
collection.add(
    ids=ids,
    documents=documents,
    embeddings=embeddings_list
)

print("Chunks stored in ChromaDB:", collection.count())
````

````python
query = "Why is Retrieval-Augmented Generation needed for large language models?"

query_embedding = embeddings.embed_query(query)

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

results
````

````python
TOP_K = 5
def retrieve_chunks(query: str, top_k: int = 5):
    # 1) Embed the user query
    query_embedding = embeddings.embed_query(query)

    # 2) Query the vector database
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k
    )

    # 3) Extract documents + ids
    retrieved_docs = []
    for i in range(len(results["documents"][0])):
        retrieved_docs.append({
            "id": results["ids"][0][i],
            "text": results["documents"][0][i]
        })

    return retrieved_docs
````

````python
question = "Why is Retrieval-Augmented Generation needed for large language models?"

retrieved = retrieve_chunks(question, top_k=TOP_K)

for r in retrieved:
    print(f"\n--- {r['id']} ---\n")
    print(r["text"][:500])
````

````python
def build_context(retrieved_chunks):
    context = ""
    for chunk in retrieved_chunks:
        context += f"[{chunk['id']}]\n{chunk['text']}\n\n"
    return context.strip()

context = build_context(retrieved)

print("===== CONTEXT SENT TO LLM =====\n")
print(context[:2000])

````

````python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
````

````python
def answer_with_rag(question: str, context: str):
    prompt = f"""
You are a research assistant.

Answer the question ONLY using the information provided in the context below.
- If the answer is not explicitly stated in the context, say: "Not found in the paper."
- Do not use any external knowledge.
- Do not make assumptions.

Context:
{context}

Question:
{question}

Answer:
"""
    response = llm.invoke(prompt)
    return response.content
````

````python
question = "Why is Retrieval-Augmented Generation needed for large language models?"

retrieved_chunks = retrieve_chunks(question, top_k=5)
context = build_context(retrieved_chunks)

answer = answer_with_rag(question, context)

print("QUESTION:\n", question)
print("\nANSWER:\n", answer)
````

````python
def generate_mermaid_diagram(question: str, context: str):
    prompt = f"""
You are a system that generates Mermaid diagrams.

RULES (follow strictly):
- Output ONLY valid Mermaid code.
- Do NOT include explanations.
- Do NOT include markdown fences like ```mermaid.
- Use ONLY the information in the context.
- If the diagram cannot be created from the context, output: NOT_FOUND

Context:
{context}

Task:
Create a Mermaid flowchart that answers the question below.

Question:
{question}

Mermaid diagram:
"""
    response = llm.invoke(prompt)
    return response.content.strip()
````

````python
question = "Explain the Retrieval-Augmented Generation pipeline"

retrieved_chunks = retrieve_chunks(question, top_k=5)
context = build_context(retrieved_chunks)

mermaid_code = generate_mermaid_diagram(question, context)

print("MERMAID OUTPUT:\n")
print(mermaid_code)
````

````python
import gradio as gr

# ---------------- Backend wrappers for UI ----------------

def rag_answer_ui(question):
    retrieved_chunks = retrieve_chunks(question, top_k=5)
    context = build_context(retrieved_chunks)
    answer = answer_with_rag(question, context)
    return answer


def rag_mermaid_ui(question):
    retrieved_chunks = retrieve_chunks(question, top_k=5)
    context = build_context(retrieved_chunks)
    mermaid_code = generate_mermaid_diagram(question, context)
    return mermaid_code


# ---------------- Gradio UI ----------------

with gr.Blocks(title="RAG Research Paper Assistant") as demo:

    gr.Markdown("## 📄 RAG Research Paper Assistant")
    gr.Markdown(
        "Ask questions grounded in the research paper. "
        "You can also generate Mermaid diagram code explaining concepts."
    )

    question_input = gr.Textbox(
        label="Enter your question",
        placeholder="Why is Retrieval-Augmented Generation needed?",
        lines=2
    )

    with gr.Row():
        answer_btn = gr.Button("🔍 Get Grounded Answer")
        diagram_btn = gr.Button("🧩 Generate Mermaid Diagram")

    answer_output = gr.Textbox(
        label="📌 Grounded Answer",
        lines=6
    )

    diagram_output = gr.Textbox(
        label="🧠 Mermaid Diagram Code",
        lines=10
    )

    answer_btn.click(
        fn=rag_answer_ui,
        inputs=question_input,
        outputs=answer_output
    )

    diagram_btn.click(
        fn=rag_mermaid_ui,
        inputs=question_input,
        outputs=diagram_output
    )

# ---------------- Launch ----------------
demo.launch()
````
